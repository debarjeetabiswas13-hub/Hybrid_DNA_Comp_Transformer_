```python
import matplotlib.pyplot as plt

# Dataset Names
datasets = [
    'BuEb','AgPh','YeMi','AeCa','HePy',
    'HaHi','EsCo','PlFa','ScPo','EnIn',
    'DrMe','OrSa','DaRe','GaGa','HoSa'
]

# Compression Time (CT)
CT = [
    10, 14.59, 14.94, 104, 118,
    391, 369.89, 121, 152, 380,
    1300, 1000, 923, 1080, 1600
]

# Decompression Time (DT)
DT = [
    0, 0, 0, 3, 2.49,
    9, 9.35, 18, 21, 65,
    58.66, 90, 140, 100, 121
]

# Create Figure
plt.figure(figsize=(12,6))

# Plot Compression Time
plt.plot(
    datasets,
    CT,
    marker='o',
    linewidth=2,
    label='Compression Time (CT)'
)

# Plot Decompression Time
plt.plot(
    datasets,
    DT,
    marker='s',
    linewidth=2,
    label='Decompression Time (DT)'
)

# Labels
plt.xlabel('Dataset', fontsize=20)
plt.ylabel('Time (s)', fontsize=20)
plt.title('HybridDNAComp Compression and Decompression Time', fontsize=14)

# X-axis
plt.xticks(rotation=45, fontsize=13)

# Custom Y-axis
plt.yticks(
    [0, 100, 300, 500, 700, 900, 1100, 1300, 1500, 1700],
    fontsize=13
)

plt.ylim(0, 1700)

# Grid
plt.grid(True, linestyle='--', alpha=0.5)

# Legend
plt.legend(fontsize=14)

# Save Figure
plt.tight_layout()
plt.savefig(
    'HybridDNA_Time_Analysis.jpeg',
    dpi=400,
    bbox_inches='tight'
)

plt.show()
```
